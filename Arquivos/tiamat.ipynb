### <a id='toc2_1_2_'></a>[Supercondutividade e Temperatura crítica (TC)](#toc0_)

Nesse notebook, trabalharemos com um problema científico que deve ser compreendido pelo leitor para que possamos seguir com a utilização de Machine Learning. Como pode ser observado pelo título, faremos a predição da temperatura crítica de supercondutores, mas a pergunta inicial é: o que são os objetos que serão estudados?

A princípio, a supercondutividade é um fenômeno físico em que certos materiais perdem completamente a resistência elétrica e expulsam campos magnéticos de seu interior quando resfriados abaixo de uma temperatura crítica específica. Esses materiais apresentam algumas características interessantes:
<br><br>

• __Resistência elétrica nula__: A corrente elétrica pode fluir por um circuito fechado de material supercondutor indefinidamente, sem perda de energia ou dissipação de calor por efeito Joule;


• __Efeito Meissner__: Os supercondutores atuam como diamagnéticos perfeitos, expulsando campos magnéticos e permitindo fenômenos como a levitação magnética.
<br><br>


Enquanto isso, a temperatura crítica, é a temperatura limite abaixo da qual um material transiciona para esse estado de supercondutividade, perdendo toda a sua resistência elétrica. Acima dessa temperatura, o material se comporta como um condutor comum (gerando calor e oferecendo resistência à corrente elétrica). Abaixo dela, ocorre uma mudança de fase termodinâmica abrupta que ativa as propriedades supercondutoras.

A temperatura crítica também apresenta características que devem ser citadas:
<br><br>

• __Ela é específica para cada material__: Cada elemento ou composto possui a sua própria \(T_{c}\). Por exemplo, o mercúrio puro se torna supercondutor a 4,2 K (-268,95 °C), enquanto cerâmicas complexas (conhecidos como supercondutores de alta temperatura) podem atingir esse estado acima de 130 K (-143 °C).


• __Dependência magnética__: A temperatura crítica não é fixa se o material estiver exposto a um campo magnético. Se o campo magnético externo for muito forte (chamado de campo crítico), ele pode destruir o estado supercondutor mesmo que o material esteja abaixo da \(T_{c}\).


• __O "Santo Graal" da Física__: Atualmente, os supercondutores práticos exigem refrigeração cara com hélio ou nitrogênio líquido. Encontrar um material com uma temperatura crítica que opere na temperatura ambiente (cerca de 293 K ou 20 °C) revolucionaria a tecnologia global, permitindo redes elétricas sem nenhuma perda de energia.
<br><br>


Com isso, os supercondutores possuem aplicações revolucionárias devido às suas propriedades. Na medicina, eles alimentam os potentes ímãs dos aparelhos de ressonância magnética, permitindo imagens diagnósticas de alta resolução. No transporte, viabilizam os trens de levitação magnética (Maglev), que flutuam sobre os trilhos e viajam a velocidades altíssimas sem atrito. No setor de energia, abrem caminho para redes de transmissão elétrica sem perdas de carga e são fundamentais nos reatores de fusão nuclear Tokamak, que usam campos magnéticos para confinar o plasma superaquecido. Por fim, na computação, são a base para o desenvolvimento de computadores quânticos e processadores ultrarrápidos.


### <a id='toc2_1_3_'></a>[Fatores físico-químicos relacionados à supercondutividade](#toc0_)

Existe um estrito conjunto de fatores físico-químicos que determinam se um material pode entrar no estado quântico macroscópico de supercondutividade, evidenciando que a interação entre a estrutura da matéria, a temperatura e as forças externas molda o comportamento do material. 

Fundamentalmente, existem três fatores físicos críticos que determinam se o material retornará (ou não) ao seu estado condutor normal:
<br><br>

• __Temperatura Crítica ($T_{c}$)__: É a temperatura abaixo da qual a agitação térmica é baixa o suficiente para permitir a atração indireta entre elétrons. Varia de quase zero absoluto (como o Tungstênio a 0,015 K) até patamares mais altos em cerâmicas complexas (cupratos);

• __Campo Magnético Crítico ($H_{c}$)__: Campos magnéticos externos destroem o estado supercondutor. Supercondutores do Tipo I possuem um único limite rígido ($H_{c}$). Já os do Tipo II toleram um estado misto (vórtices quantizados) entre um campo inferior ($H_{c1}$) e superior ($H_{c2}$), tornando-os úteis para gerar campos magnéticos massivos.

• __Densidade de Corrente Crítica ($J_{c}$)__: A quantidade máxima de corrente elétrica por unidade de área que o material transporta sem resistência. Correntes que excedem esse limite geram campos magnéticos induzidos que destroem a supercondutividade.

<br><br>


Além disso, a composição química e a organização atômica influenciam diretamente os mecanismos quânticos que estabilizam o fenômeno:

<br><br>

• __Aceleração Fônon-Elétron (Rede Cristalina)__: Na teoria convencional (BCS), a passagem de um elétron deforma localmente os íons positivos da rede cristalina. Essa distorção (vibração molecular ou fônon) cria uma região temporária de carga positiva que atrai um segundo elétron, superando a repulsão natural de Coulomb. Esse par ligado chama-se __Par de Cooper__;


• __Anisotropia e Planos de Óxido de Cobre ($\text{CuO}_{2}$)__: Em supercondutores de alta temperatura (cupratos), a condução ocorre de forma bidimensional ao longo de planos de óxido de cobre intercalados por camadas isolantes. A geometria e a distância entre esses planos ditam a eficácia da supercondutividade;


• __Dopagem Química__: Pequenas alterações estequiométricas mudam drasticamente o comportamento do material. A introdução controlada de átomos (como oxigênio em cerâmicas ou dopantes químicos em fulerenos) ajusta a densidade de portadores de carga até atingir um ponto ideal de transição eletrônica.


• __Impurezas Magnéticas__: Elementos com propriedades magnéticas fortes (como manganês ou ferro) quebram a simetria de spin dos Pares de Cooper, atuando como venenos químicos que destroem ou reduzem severamente a $T_{c}$ do material.


<br><br>

Por fim, a aplicação de __pressões mecânicas extremas__ também pode alterar a supercondutividade. Ao comprimir a rede cristalina, as frequências de vibração dos átomos (fônons) são alteradas.

### <a id='toc2_1_4_'></a>[Por que utilizar Machine Learning?](#toc0_)

Como vimos anteriormente, a temperatura crítica ($T_c$) é o maior fator limitante para a aplicação em grande escala de materiais supercondutores. Além desse problema, tem-se também que nenhuma teoria científica consegue prever o valor de $T_c$ de forma universal. Apesar da existência de teorias importantes como a __BCS__, elas ainda falham em explicar e prever as propriedades de alguns tipos de supercondutores, como, por exemplo, os supercondutores de alta temperatura.

Assim, o aprendizado de máquina pode ser uma ferramenta interessante para contornar este problema. Isso porque, ao utilizarmos um dataset com features que contêm boa parte das propriedades físico-químicas dos supercondutores, seria possível induzir modelos para prever $T_c$, explorando uma possível solução.

### <a id='toc2_1_5_'></a>[Objetivos e hipóteses](#toc0_)

O objetivo geral desse notebook é induzir modelos de Machine Learning capazes de prever a temperatura crítica de supercondutores a partir das features do dataset. 

Nesse sentido, de forma mais específica, buscamos checar o desempenho dos diferentes modelos induzidos, analisando quais deles puderam captar de forma satisfatória os padrões do dataset e também pretendemos investigar em que medida características derivadas da composição química de materiais supercondutores permitem prever sua temperatura crítica $T_c$, e como a seleção de atributos e a escolha do algoritmo influenciam a qualidade das previsões.

Inicialmente, existem hipóteses científicas preliminares que guiarão o projeto:

1. Modelos baseados em árvores e ensembles apresentam melhor capacidade preditiva do que regressões lineares, devido às relações complexas entre composição e propriedades supercondutoras.

2. Um subconjunto selecionado dos 81 descritores pode alcançar desempenho próximo ao conjunto completo, reduzindo complexidade e facilitando interpretação.

3. Determinados grupos de descritores, como estatísticas de elétrons de valência ou propriedades atômicas, apresentam maior contribuição preditiva do que outros.


## <a id='toc2_2_'></a>[Apresentação e análise exploratória dos dados](#toc0_)

### <a id='toc2_2_1_'></a>[O dataset](#toc0_)

O dataset que será utilizado no projeto provém do repositório __UCI Machine Learning Repository__, um repositório de datasets da Universidade da California (Irvine). Ele é nomeado como "Supercondutivity Data" e contém 21263 dados reais de supercondutores, com 81 features e o valor da temperatura crítica (em Kelvin) como target.

O 'criador' desse dataset categorizado como um problema de regressão é o professor Kam Hamidieh, que possui um artigo de 2018 (referência [1]) que utiliza ML para investigar essa mesma base de dados, sendo assim a principal referência para a produção desse projeto.

Kam Hamidieh deixa claro em seu artigo que ele importou dos dados dos supercondutores do database Superconducting Material Database. Este banco de dados foi inicialmente produzido pelo Instituto Nacional de Ciência dos Materiais (NIMS - National Institute for Materials Science) — a principal agência nacional de pesquisa e desenvolvimento do Japão dedicada exclusivamente à ciência e tecnologia de materiais. Dessa forma, ele importou esses dados e produziu o dataset.

Por isso, a página diz que não há valores faltantes no dataset. E também diz que a base de dados está separada em dois arquivos: 

1. __"train.csv"__: possui 81 features extraídas de cada um dos supercondutores e a temperatura crítica;

2. __"unique_m.csv"__: contém a fórmula química de cada um dos 21263 supercondutores.

A partir dessas propriedades, foram criados descritores estatísticos: médias, médias ponderadas, médias geométricas, entropias, amplitudes, desvios padrão, entre outros... No artigo, o autor dá um exemplo de como as features foram extraídas do banco de dados.

Considerando o supercondutor $\text{Re}_{7}{Zr}_{1}$ com $T_c$ = 6.7 K :

A condutividade térmica do Rênio e do zircônia são $t_1$ = 48 e $t_2$ = 23 $\frac{W}{(m\times K)}$, respectivamente. 

Dessa forma, as razões entre os elementos no material são utilizadas para definir características:

$$
p_1 = \frac{6}{6+1} = \frac{6}{7},
\qquad
p_2 = \frac{1}{6+1} = \frac{1}{7}.

$$

As frações das condutividades térmicas totais também são utilizadas:

$$
w_1 = \frac{t_1}{t_1+t_2}
= \frac{48}{48+23}
= \frac{48}{71},
$$

$$
w_2 = \frac{t_2}{t_1+t_2}
= \frac{23}{48+23}
= \frac{23}{71}.
$$

Precisamos de alguns valores intermediários baseados nas Eqs.~(1) e (2):

$$
A =
\frac{p_1w_1}{p_1w_1+p_2w_2}
\approx 0.926,
$$

$$
B =
\frac{p_2w_2}{p_1w_1+p_2w_2}
\approx 0.074.
$$

Após obtermos os valores de $p_1$, $p_2$, $w_1$, $w_2$, $A$ e $B$, torna-se possível extrair 10 features a partir das condutividades térmicas de Rênio e Zircônio, conforme apresentado na Tabela 2.

<br><br>

__Tabela 2__: Resumo das features extraídas a partir da fórmula química de um material. A última coluna apresenta os valores obtidos para a condutividade térmica do composto $\text{Re}_{7}{Zr}_{1}$, considerando $t_1 = 48$ e $t_2 = 23$   $\frac{W}{(m\times K)}$,.

| Característica | Descrição | Fórmula | Valor de exemplo |
|---------------|-----------|----------|------------------|
| Média | Média aritmética das propriedades dos elementos | μ = (t₁ + t₂)/2 | 35,5 |
| Média ponderada | Média ponderada pelas proporções estequiométricas | ν = p₁t₁ + p₂t₂ | 44,43 |
| Média geométrica | Média geométrica das propriedades | (t₁t₂)¹ᐟ² | 33,23 |
| Média geométrica ponderada | Média geométrica ponderada pelas proporções estequiométricas | (t₁)ᵖ¹(t₂)ᵖ² | 43,21 |
| Entropia | Entropia baseada nas frações das propriedades | −w₁ln(w₁) − w₂ln(w₂) | 0,63 |
| Entropia ponderada | Entropia baseada nos pesos A e B | −Aln(A) − Bln(B) | 0,26 |
| Amplitude | Diferença entre os valores máximo e mínimo | t₁ − t₂ (t₁ > t₂) | 25 |

| Desvio padrão | Dispersão dos valores em torno da média | [½((t₁ − μ)² + (t₂ − μ)²)]¹ᐟ² | 12,5 |
| Desvio padrão ponderado | Dispersão ponderada em torno da média ponderada | [p₁(t₁ − ν)² + p₂(t₂ − ν)²]¹ᐟ² | 8,75 |

O resultado representa uma propriedade da composição do material, expressa numericamente. Isso permite que os algoritmos de ML utilizem informações químicas sem precisar interpretar diretamente a fórmula molecular.


Um ponto físico importante: a média ponderada não informa como os átomos estão organizados no cristal. Materiais com a mesma composição, mas estruturas ou condições físicas diferentes, podem apresentar propriedades distintas.

Utilizaremos a correlação de Pearson:

$$
r_{X,Y}=
\frac{\operatorname{Cov}(X,Y)}
{\sigma_X\sigma_Y}
$$

Seu valor está entre $-1$ e $1$. Valores próximos aos extremos indicam associações lineares fortes, enquanto valores próximos de zero indicam pouca associação linear, mas não necessariamente ausência de relação.

```python

correlacoes = (
    df.corr(numeric_only=True)["critical_temp"]
    .drop("critical_temp")
)


top_atributos = (
    correlacoes.abs()
    .sort_values(ascending=False)
    .head(15)
    .index
)

top_correlacoes = correlacoes.loc[top_atributos].sort_values()

fig, ax = plt.subplots(figsize=(10, 6))

ax.barh(
    top_correlacoes.index,
    top_correlacoes.values,
    color="steelblue"
)

ax.axvline(0, color="black", linewidth=0.8)
ax.set_title("15 atributos mais correlacionados com Tc")
ax.set_xlabel("Coeficiente de correlação de Pearson")
ax.set_ylabel("Atributos físico-químicos")
ax.set_xlim(-1, 1)
ax.grid(axis="x", alpha=0.2)

plt.tight_layout()
plt.show()
```

### <a id='toc2_3_1_'></a>[Separação entre treinamento e teste](#toc0_)

Um erro metodológico grave seria treinar um modelo sobre todos os dados e depois avaliar seu desempenho nesses mesmos registros. Isso produziria uma estimativa excessivamente otimista da capacidade de generalização. 


A solução é reservar um conjunto de teste, que permanecerá separado durante a otimização e a comparação dos modelos. Dessa forma, teremos:

- **Treinamento**: Usado na validação cruzada, escolha dos modelos e otimização com Optuna;

- **Teste**: Reservado para avaliação final, após as decisões de modelagem.

#### <a id='toc2_3_1_1_'></a>[Uma particularidade do nosso dataset](#toc0_)

No Capítulo 2, examinamos a possibilidade de existirem composições químicas repetidas. Essa particularidade gera um problema: se registros de uma mesma composição aparecerem nos conjuntos de treino e teste, o modelo poderá reconhecer características de um material que já encontrou durante o treinamento.

Uma solução para isso seria adotarmos como protocolo principal uma divisão agrupada pela composição química.

Essa escolha é mais exigente do que uma divisão aleatória convencional, pois impede que a mesma fórmula registrada apareça simultaneamente no treinamento e no teste. Entretanto, não elimina todas as semelhanças entre materiais diferentes.

Antes de executar a divisão, também precisamos verificar como representaremos fórmulas equivalentes, por exemplo fórmulas escritas em ordens diferentes.

Partiremos dos DataFrames df e df_composicoes já carregados no Capítulo 2.

Além disso, vale a pena citar que utilizaremos uma SEED fixa para garantir a reprodutibilidade dos resultados.

```python
from sklearn.model_selection import GroupShuffleSplit

SEED = 210906

X = df.drop(columns="critical_temp")
y = df["critical_temp"]

assert len(df) == len(df_composicoes)

assert np.allclose(
    y.to_numpy(),
    df_composicoes["critical_temp"].to_numpy()
)


grupos = df_composicoes["material"].astype(str)

divisor = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=SEED
)

indice_treino, indice_teste = next(
    divisor.split(X, y, groups=grupos)
)

X_treino = X.iloc[indice_treino].copy()
X_teste = X.iloc[indice_teste].copy()

y_treino = y.iloc[indice_treino].copy()
y_teste = y.iloc[indice_teste].copy()

grupos_treino = grupos.iloc[indice_treino].copy()
grupos_teste = grupos.iloc[indice_teste].copy()

print("Treino:", X_treino.shape)
print("Teste:", X_teste.shape)


intersecao = (
    set(grupos_treino) & set(grupos_teste)
)

assert len(intersecao) == 0
print(f"Composições compartilhadas: {len(intersecao)}")
```

Dessa forma, agora temos uma divisão treino-teste controlada por grupos, onde o grupo é a fórmula química do material. A ideia é evitar que o modelo veja a mesma composição química tanto no treino quanto no teste, o que poderia gerar uma avaliação artificialmente otimista.

Usando GroupShuffleSplit, a pergunta que você faz ao modelo muda de:

"Você consegue prever uma amostra parecida com algo que já viu?"

para:

"Você consegue prever a temperatura crítica de uma composição química inteiramente nova?"

Essa segunda avaliação é muito mais rigorosa e costuma refletir melhor a capacidade real de generalização dos modelos.

### <a id='toc2_3_2_'></a>[Validação cruzada](#toc0_)

Após separarmos os dados, surge outra questão de como escolheremos os melhores hiperparâmetros e compararemos os algoritmos. Para isso, utilizaremos validação cruzada.

Na validação cruzada K-Fold, os dados de treinamento são divididos em $K$ subconjuntos. A cada iteração, um subconjunto é utilizado para validação e os demais para treinamento. Um exemplo visual desse processo pode ser visualizado abaixo:

Para esse notebook, utilizaremos GroupKFold, pois queremos manter cada composição química inteiramente em um único fold de validação.

```python

from sklearn.model_selection import GroupKFold

cv = GroupKFold(n_splits=5)

folds = list(
    cv.split(
        X_treino,
        y_treino,
        groups=grupos_treino
    )
)

for i, (indice_treino, indice_validacao) in enumerate(folds, start=1):
    composicoes_treino = set(
        grupos_treino.iloc[indice_treino]
    )
    composicoes_validacao = set(
        grupos_treino.iloc[indice_validacao]
    )

    assert composicoes_treino.isdisjoint(composicoes_validacao)

    print(
        f"Fold {i}: "
        f"{len(indice_treino)} treino | "
        f"{len(indice_validacao)} validação"
    )
```

Esses mesmos folds serão utilizados para comparar os cinco algoritmos, o que torna a avaliação mais consistente.

### <a id='toc2_3_3_'></a>[Processamento dos dados e construção dos pipelines](#toc0_)

Para a construção dos pipelines e indução dos diferentes modelos, utilizaremos a biblioteca do scikit-learn. 

Scikit-learn é uma biblioteca de aprendizado de máquina de código aberto que suporta aprendizagem supervisionada e não supervisionada. Também fornece várias ferramentas para encaixe do modelo, pré-processamento de dados, seleção de modelos, avaliação do modelo, e muitas outras utilidades.



Os atributos utilizados neste trabalho representam diferentes propriedades físico-químicas dos materiais supercondutores. Por essa razão, apresentam distintas unidades, distribuições e ordens de grandeza.

Essa heterogeneidade deve ser considerada durante a preparação dos dados, uma vez que alguns algoritmos de aprendizado de máquina são sensíveis à escala das variáveis de entrada.

#### <a id='toc2_3_3_1_'></a>[Padronização dos atributos](#toc0_)

A padronização consiste em transformar os valores de um atributo tomando como referência sua média e seu desvio-padrão. Na padronização Z-score, essa transformação é definida por:

$$
z_i = \frac{x_i-\mu}{\sigma}
\tag{}
$$

em que $x_i$ é o valor original, $\mu$ é a média e $\sigma$ é o desvio-padrão do atributo, calculados sobre os dados utilizados para ajustar a transformação.

Esse procedimento é especialmente relevante para algoritmos baseados em distâncias e para modelos lineares regularizados.

Entretanto, os algoritmos baseados em árvores de decisão são geralmente pouco sensíveis a transformações lineares da escala dos atributos, uma vez que constroem suas previsões a partir de divisões sucessivas dos valores das variáveis.

Dessa forma, adotaremos a padronização para a Ridge Regression, enquanto Random Forest, Extra Trees, HistGradientBoosting e XGBoost utilizarão inicialmente os atributos em suas escalas originais.

### <a id='toc2_3_4_'></a>[Construção de pipelines e prevenção de vazamento de dados](#toc0_)

Para garantir a consistência das transformações e evitar vazamento de dados (data leakage), utilizaremos pipelines do scikit-learn.

Um pipeline organiza sequencialmente as operações de pré-processamento e treinamento. Durante a validação cruzada, as transformações que dependem dos dados, como a padronização e a seleção supervisionada de atributos, são ajustadas exclusivamente com os registros de treinamento de cada fold.

Isso impede que informações provenientes dos conjuntos de validação ou teste influenciem indevidamente o treinamento.

Os cinco algoritmos serão organizados em pipelines, permitindo padronizar o procedimento experimental e integrar posteriormente a otimização de hiperparâmetros.

#### <a id='toc2_3_4_1_'></a>[Seleção de atributos](#toc0_)

O conjunto de dados utilizado possui 81 atributos preditivos. Entretanto, nem todos necessariamente fornecem informações independentes ou igualmente úteis para estimar a temperatura crítica.

### <a id='toc2_3_5_'></a>[Definição das métricas de avaliação](#toc0_)

As métricas de avaliação fazem o trabalho de quantificar o quão próximo o modelo chega dos valores reais, utilizando diferentes cálculos para medir o erro entre as previsões e os resultados já observados. Existem diversas métricas para medir o desempenho de modelos de regressão,  mas, para esse notebook, utilizaremos três. Sendo elas: a métrica do erro médio absoluto (MAE), a métrica da raiz do erro quadrático médio (RMSE) e erro percentual absoluto médio com arco-tangente (MAAPE), visto que as duas primeiras são intuitivas e fornecem a informação necessária sobre o desempenho, enquanto a última será utilizada como métrica complementar. 

Essa combinação é interessante, pois o MAE e o RMSE medem coisas diferentes e um pode complementar o outro, já que o RMSE é mais sensível à outliers. Nesse contexto, o MAAPE permite investigar aspectos complementares dos erros produzidos pelos algoritmos. 

#### <a id='toc2_3_5_1_'></a>[Erro médio absoluto (MAE)](#toc0_)

O erro médio absoluto é uma métrica que indica a __magnitude média dos erros em um conjunto de previsões__, calculando a média das diferenças absolutas entre os valores reais e os valores previstos pelo modelo. Uma característica importante do MAE é que grandes desvios não são penalizados de forma desproporcional, então a presença de outliers não tem impacto tão grande no resultado, o que pode ser uma vantagem. Por outro lado, em casos onde o modelo apresenta erros por grandes margens, o resultado do MAE pode não refletir bem o impacto desses desvios. Sua fórmula é dada por:

$$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$

em que:

-   $y_i$ é o target do exemplo $i$ (valor que conhecemos, pode ser chamado de &ldquo;$y$ verdadeiro&rdquo; ou &ldquo;$y$ true&rdquo; na literatura);

-   $\hat{y}_i$ é o valor previsto para o exemplo $i$;

-   e $n$ é o número de exemplos do conjunto de dados que estamos computando a métrica.

#### <a id='toc2_3_5_2_'></a>[Métrica da raiz do erro quadrático médio (RMSE)](#toc0_)

O RMSE (Root Mean Squared Error) mede a magnitude dos erros de previsão, __calculando a raiz quadrada da média dos erros quadráticos__. Essa métrica é mais sensível a grandes desvios, pois o erro é elevado ao quadrado, o que significa que grandes diferenças entre o valor previsto e o valor real terão um grande impacto no resultado. O RMSE é ideal para cenários onde grandes desvios são considerados prejudiciais e devem ser minimizados.

Sua fórmula está expressa abaixo:

$$
\mathrm{RMSE} = \sqrt{\mathrm{MSE}} = \sqrt{\sum_{i=1}^{n} \frac{(y_i - \hat{y}_i)^2}{n}}.
$$


em que:

-   $y_i$ é o target do exemplo $i$;

-   $\hat{y}_i$ é o valor previsto para o exemplo $i$;

-   e $n$ é o número de exemplos do conjunto de dados que estamos computando a métrica.

#### <a id='toc2_3_5_3_'></a>[Mean Arctangent Absolute Percentage Error (MAAPE)](#toc0_)

A MAAPE foi proposta por Kim e Kim (2016) como alternativa ao MAPE, buscando limitar a influência de erros relativos muito elevados.

Sua expressão é:

$$
MAAPE = \frac{1}{n}\sum_{i=1}^{n}\arctan\left(\left|\frac{y_i-\hat{y}_i}{y_i}\right|\right)
$$

A função arco-tangente limita cada contribuição ao intervalo de 0 a $\pi/2$, permitindo analisar erros relativos sem que valores percentuais extremos cresçam indefinidamente.

Ele consegue olhar o MAPE por outro ângulo, preservando a sua filosofia, mas superando uma limitação essencial do MAPE: a divisão por zero. Isso porque, a métrica MAPE fica com valor indefinido ou tende ao infinito tende ao infinito se algum valor real for próximo de zero. Neste trabalho, temos valores de $T_c$ bem próximos de zero, por isso, o MAAPE é uma métrica mais interessante.

Além disso, se obtivermos valores observados iguais a zero, adotaremos a extensão por limite: a contribuição será zero quando a previsão também for zero e $\pi/2$ quando a previsão for diferente de zero.

O resultado será apresentado em radianos. Nesse sentido, MAAPE = 0: É a previsão perfeita. Significa que todas as previsões ($f_{t}$) foram exatamente iguais aos valores reais ($y_{t}$), de forma que o valor do MAAPE sempre estará no intervalo $[0, \pi/2)$, ou seja, entre 0 e aproximadamente 1,5708.

A MAAPE será utilizada como métrica complementar, enquanto o MAE permanecerá como critério principal de seleção dos hiperparâmetros.

#### <a id='toc2_3_5_4_'></a>[Definição das métricas (código)](#toc0_)

A partir de tudo que foi exposto, vamos definir as métricas na célula de código abaixo.

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)


def calcular_maape(y_real, y_previsto):
    y_real = np.asarray(y_real, dtype=float)
    y_previsto = np.asarray(y_previsto, dtype=float)

    erro = np.abs(y_real - y_previsto)
    termos = np.empty_like(erro)

    mascara = y_real != 0

    termos[mascara] = np.arctan(
        erro[mascara] / np.abs(y_real[mascara])
    )

    termos[~mascara] = np.where(
        erro[~mascara] == 0,
        0.0,
        np.pi / 2
    )

    return float(np.mean(termos))


def calcular_metricas(y_real, y_previsto):
    return {
        "MAE (K)": mean_absolute_error(
            y_real, y_previsto
        ),
        "RMSE (K)": np.sqrt(
            mean_squared_error(y_real, y_previsto)
        ),
        "MAAPE (rad)": calcular_maape(
            y_real, y_previsto
        )
    }
```

### <a id='toc2_4_5_'></a>[Preparação das avaliações comum](#toc0_)

Antes de treinar os modelos, vamos criar um procedimento que calcule as mesmas métricas para todos eles.

```python
from sklearn.model_selection import cross_validate
from sklearn.metrics import make_scorer

scorers = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "MAAPE": make_scorer(
        calcular_maape,
        greater_is_better=False
    )
}


def avaliar_modelo(nome, modelo, X, y, folds):
    """
    Avalia um modelo nos folds definidos e
    retorna as métricas individuais.
    """

    resultado = cross_validate(
        estimator=modelo,
        X=X,
        y=y,
        cv=folds,
        scoring=scorers,
        n_jobs=1,
        error_score="raise"
    )

    return pd.DataFrame({
        "Modelo": nome,
        "Fold": range(1, len(folds) + 1),
        "MAE (K)": -resultado["test_MAE"],
        "RMSE (K)": -resultado["test_RMSE"],
        "MAAPE (rad)": -resultado["test_MAAPE"],
        "Tempo de ajuste (s)": resultado["fit_time"]
    })
```

Essa função não utiliza o conjunto de teste reservado e não altera os folds.

Antes de implementar, precisamos distinguir três conceitos: seleção de atributos, extração de atributos e importância de atributos. Neste projeto, priorizaremos seleção de atributos, porque ela preserva os significados físico-químicos dos descritores originais.

Inicialmente, será utilizado o método SelectKBest, que seleciona os atributos de acordo com uma função de pontuação estatística. Como critério, será considerada a função f_regression, baseada na associação linear individual entre cada descritor e a variável-alvo.

O número de atributos selecionados será tratado como um hiperparâmetro e, posteriormente, otimizado com Optuna. Também será considerada a possibilidade de utilizar todos os atributos, permitindo avaliar se a seleção efetivamente beneficia o desempenho preditivo.

### <a id='toc2_4_7_'></a>[Adicionando a seleção aos cinco pipelines](#toc0_)

Utilizaremos os pipelines criados nas seções anteriores e a função clone, que gera novos estimadores com as mesmas configurações, mas ainda não treinados.

```python

from sklearn.base import clone
from sklearn.feature_selection import SelectKBest, f_regression

def construir_pipeline_com_selecao(pipeline_base, k="all"):
    """
    Adiciona seleção de atributos antes das demais etapas
    de um pipeline existente, sem ajustar os modelos.
    """
    etapas = [
        ("selecao", SelectKBest(
            score_func=f_regression,
            k=k
        ))
    ]

    etapas.extend(clone(pipeline_base).steps)

    return Pipeline(etapas)
```

Observe que k="all" significa utilizar todos os 81 atributos, sem eliminá-los. Essa opção é oficialmente suportada pelo SelectKBest e permite comparar modelos com e sem redução do conjunto de variáveis. Em caso de curiosidade, leia:  [SelectKBest](https://scikit-learn.org/stable/modules/generated/sklearn.feature_selection.SelectKBest.html).

## <a id='toc2_5_'></a>[Otimização dos hiperparâmetros](#toc0_)

Nesta etapa, começaremos a otimizar os algoritmos que implementamos anteriormente. A ideia é substituir escolhas arbitrárias de hiperparâmetros por uma busca sistemática, avaliada por validação cruzada.

### <a id='toc2_5_1_'></a>[Fundamentação rápida](#toc0_)

#### <a id='toc2_5_1_1_'></a>[O que são hiperparâmetros?](#toc0_)

Precisamos distinguir dois conceitos.

Parâmetros são valores aprendidos durante o treinamento. Na Ridge Regression, por exemplo, são os coeficientes associados aos atributos e o intercepto.

Hiperparâmetros são configurações que controlam o treinamento ou a estrutura do modelo. Na Ridge, um exemplo é alpha, responsável por controlar a intensidade da regularização.

O objetivo da otimização é encontrar uma configuração de hiperparâmetros que produza bom desempenho em dados não utilizados no ajuste dos parâmetros do modelo.

#### <a id='toc2_5_1_2_'></a>[Como o Optuna funciona?](#toc0_)

O Optuna organiza a otimização em dois conceitos principais:

- Study: representa um estudo de otimização inteiro.

- Trial: representa a avaliação de uma configuração de hiperparâmetros.

Cada trial propõe uma configuração, treina e avalia um modelo e retorna um valor numérico para o Optuna.

Em nosso caso, esse valor será o MAE médio obtido na validação cruzada agrupada.

Ele escolhe os hiperparâmetros de diversas formas possíveis, mas, no nosso caso, utilizaremos o TPESampler, baseado no algoritmo Tree-structured Parzen Estimator (TPE).

Diferentemente de uma busca em grade, que avalia combinações previamente definidas, o TPE utiliza os resultados anteriores para orientar a amostragem de novas configurações promissoras.

Isso não garante encontrar o ótimo global, mas pode tornar a busca mais eficiente em comparação com uma exploração exaustiva do espaço.


### <a id='toc2_5_2_'></a>[Otimização, de fato](#toc0_)

#### <a id='toc2_5_2_1_'></a>[Ridge Regression](#toc0_)

Começaremos pela Ridge porque precisamos otimizar apenas um hiperparâmetro principal do estimador, além da seleção de atributos.

A Ridge procura minimizar:

$$
J(\boldsymbol{\beta})=
\underbrace{\sum_{i=1}^{n}(y_i-\hat y_i)^2}_{\text{Erro de ajuste}}
+
\underbrace{\alpha\sum_{j=1}^{p}\beta_j^2}_{\text{Regularização L2}}
$$

O hiperparâmetro $\alpha$ controla a intensidade da penalização.

O hiperparâmetro $\alpha$ controla a intensidade da penalização.
- Valores pequenos de $\alpha$ permitem coeficientes menos penalizados.
- Valores grandes de $\alpha$ tendem a reduzir a magnitude dos coeficientes.
- Regularização excessiva pode produzir underfitting, enquanto regularização insuficiente pode aumentar a sensibilidade a ruído.

__Observe uma distinção__: a Ridge ajusta seus coeficientes minimizando uma função baseada em erros quadráticos, mas nós escolheremos seus hiperparâmetros minimizando o MAE de validação cruzada. Não há contradição nisso.

##### <a id='toc2_5_2_1_1_'></a>[Definição do espaço de busca](#toc0_)

| Hiperparâmetro | Intervalo | Justificativa |
|---------------|-----------|---------------|
| `alpha` | \(10^{-4}\) a \(10^4\) | Explorar regularização fraca e forte |
| `usar_selecao` | Verdadeiro ou falso | Investigar o benefício de selecionar atributos |
| `k` | 10 a 80, quando há seleção | Explorar diferentes tamanhos de subconjuntos |

Vamos utilizar uma escala logarítmica para alpha, porque queremos explorar várias ordens de grandeza, não apenas números distribuídos uniformemente.

O intervalo é uma escolha experimental inicial, não um intervalo universalmente ótimo para Ridge. Como utilizamos StandardScaler, a interpretação do efeito de alpha será mais consistente entre atributos com escalas originalmente diferentes.

```python
import optuna

from pathlib import Path
from sklearn.model_selection import cross_validate

N_TRIALS = 100

assert len(X_treino) == len(y_treino)
assert X_treino.shape[1] == 81
assert len(folds) == 5

pasta_optuna = Path("results/optuna")
pasta_optuna.mkdir(parents=True, exist_ok=True)

print("Ambiente preparado para otimização.")
```

##### <a id='toc2_5_2_1_3_'></a>[Função objetivo](#toc0_)

Vamos construir a função objetivo nesse código. Esta é a parte mais importante da implementação.

A função objetivo recebe um trial, solicita os hiperparâmetros ao Optuna, constrói o pipeline e calcula o MAE médio por validação cruzada.

```python

def objetivo_ridge(trial):

    
    alpha = trial.suggest_float(
        "alpha",
        1e-4,
        1e4,
        log=True
    )

    
    usar_selecao = trial.suggest_categorical(
        "usar_selecao",
        [True, False]
    )

    if usar_selecao:
        k = trial.suggest_int("k", 10, 80)
    else:
        k = "all"

    
    pipeline = construir_pipeline_com_selecao(
        pipelines["Ridge"],
        k=k
    )

    print(type(pipeline))
    print(pipeline)
    
    pipeline.set_params(
        modelo__alpha=alpha
    )

   
    resultado = cross_validate(
        estimator=pipeline,
        X=X_treino,
        y=y_treino,
        cv=folds,
        scoring="neg_mean_absolute_error",
        n_jobs=1,
        error_score="raise"
    )

    
    erros = -resultado["test_score"]

    
    trial.set_user_attr(
        "mae_desvio",
        float(np.std(erros, ddof=1))
    )

    trial.set_user_attr(
        "mae_por_fold",
        erros.tolist()
    )

    return float(np.mean(erros))
```

trial.suggest_float() permite que o Optuna escolha um valor real dentro do intervalo especificado. Já trial.suggest_int() seleciona um inteiro.

O pipeline.set_params(modelo__alpha=alpha) configura o parâmetro alpha do estimador chamado modelo dentro do pipeline.

A função cross_validate() ajusta uma nova cópia do pipeline em cada fold, o que garante que a seleção dos atributos e a padronização sejam ajustadas exclusivamente aos registros de treinamento correspondentes.

Por fim, a função retorna apenas um valor: o MAE médio dos cinco folds. É esse valor que o Optuna tentará minimizar.

#### <a id='toc2_5_2_3_'></a>[Definição dos espaços e busca dos hiperparâmetros](#toc0_)

A otimização dos algoritmos de regressão foi realizada por meio da investigação de diferentes combinações de hiperparâmetros. Para cada algoritmo, foi definido um **espaço de busca**, isto é, um conjunto de valores ou intervalos que o Optuna pode explorar durante a otimização. Os intervalos foram escolhidos de modo a permitir a avaliação de modelos com diferentes níveis de complexidade, sem tornar o espaço de busca excessivamente amplo.

O procedimento utilizou o algoritmo *Tree-structured Parzen Estimator* (TPE), com a semente aleatória 210906, tendo como objetivo minimizar o MAE médio dos cinco folds da validação cruzada.

Além dos hiperparâmetros próprios de cada algoritmo, foi investigada a utilização de um procedimento de seleção de atributos.

#### <a id='toc2_5_2_4_'></a>[Seleção de atributos](#toc0_)

O processo de seleção foi incorporado ao espaço de busca dos quatro algoritmos, permitindo que o Optuna investigasse tanto o uso dos 81 atributos originais quanto a utilização de um subconjunto deles.

| Hiperparâmetro | Valores investigados | Descrição |
|---|---|---|
| `usar_selecao` | `True` ou `False` | Determina se será realizada seleção de atributos. |
| `k` | Inteiros de 10 a 80 | Número de atributos preservados quando a seleção está ativada. |

A seleção é realizada por meio do método `SelectKBest`, utilizando a função `f_regression`. Essa função avalia a associação linear individual entre cada atributo e a temperatura crítica.

Quando `usar_selecao=False`, todos os atributos são preservados e o hiperparâmetro `k` não é investigado naquele trial.

A seleção ocorre dentro do pipeline e é ajustada independentemente em cada fold. Dessa maneira, os registros de validação não participam da escolha dos atributos, evitando vazamento de informações.

Embora o procedimento permita investigar a influência da dimensionalidade, sua natureza univariada e linear constitui uma limitação, especialmente em problemas nos quais relações não lineares e interações entre propriedades químicas podem ser relevantes.

#### <a id='toc2_5_2_5_'></a>[Random Forest](#toc0_)

A Random Forest combina múltiplas árvores de decisão construídas a partir de amostras aleatórias dos dados. Na regressão, a previsão final corresponde à média das previsões individuais das árvores.

Seus hiperparâmetros controlam principalmente a quantidade de árvores, a complexidade das divisões e a aleatoriedade introduzida no treinamento.

| Hiperparâmetro | Espaço de busca | Função |
|---|---|---|
| `n_estimators` | 100 a 400, passo 50 | Determina o número de árvores da floresta. |
| `max_depth` | `None`, 5, 10, 15, 20, 25 ou 30 | Limita a profundidade máxima das árvores. |
| `min_samples_split` | Inteiros de 2 a 20 | Define a quantidade mínima de amostras necessária para dividir um nó. |
| `min_samples_leaf` | Inteiros de 1 a 10 | Estabelece o número mínimo de amostras em cada folha. |
| `max_features` | 0,3 a 1,0 | Controla a fração de atributos considerada na busca por cada divisão. |
| `max_samples` | 0,6 a 1,0 | Controla a fração de registros amostrada para construir cada árvore. |

O hiperparâmetro `n_estimators` influencia a quantidade de árvores utilizadas na agregação. O aumento desse valor pode tornar a previsão mais estável, embora também eleve o custo computacional.

Por sua vez, `max_depth`, `min_samples_split` e `min_samples_leaf` controlam a complexidade das árvores individuais. Árvores excessivamente profundas podem representar relações muito específicas dos dados de treinamento, enquanto restrições excessivas podem impedir a aprendizagem de padrões relevantes.

Os hiperparâmetros `max_features` e `max_samples` controlam aspectos da aleatoriedade utilizada na construção da floresta. Sua investigação permite avaliar diferentes equilíbrios entre a diversidade dos estimadores e a quantidade de informação disponível para cada árvore.

#### <a id='toc2_5_2_6_'></a>[Extra Trees](#toc0_)

O algoritmo *Extremely Randomized Trees* (Extra Trees) também combina múltiplas árvores de decisão. Sua principal distinção em relação à Random Forest está na maior aleatoriedade introduzida na escolha dos pontos de divisão dos atributos.

No presente experimento, foram otimizados os seguintes hiperparâmetros:

| Hiperparâmetro | Espaço de busca | Função |
|---|---|---|
| `n_estimators` | 100 a 400, passo 50 | Determina a quantidade de árvores. |
| `max_depth` | `None`, 5, 10, 15, 20, 25 ou 30 | Controla a profundidade máxima das árvores. |
| `min_samples_split` | Inteiros de 2 a 20 | Define o número mínimo de amostras para dividir um nó. |
| `min_samples_leaf` | Inteiros de 1 a 10 | Determina o número mínimo de amostras por folha. |
| `max_features` | 0,3 a 1,0 | Controla a fração de atributos considerada em cada divisão. |

Os espaços de busca foram mantidos semelhantes aos da Random Forest, permitindo investigar a influência de hiperparâmetros equivalentes nos dois algoritmos.

Entretanto, diferentemente da configuração adotada para a Random Forest, o Extra Trees utiliza o comportamento padrão de `bootstrap=False`, construindo suas árvores com o conjunto de treinamento disponível, sem realizar amostragem bootstrap dos registros.

Essa diferença, associada à aleatorização dos pontos de divisão, torna interessante comparar posteriormente os resultados dos dois algoritmos.

#### <a id='toc2_5_2_7_'></a>[HistGradientBoosting](#toc0_)

O HistGradientBoosting é um algoritmo de *gradient boosting* que constrói árvores sequencialmente, buscando corrigir os erros do conjunto de estimadores anterior.

Sua implementação utiliza uma representação dos atributos em intervalos discretizados (*histogramas*), permitindo acelerar a busca por divisões durante o treinamento.

Foram investigados os seguintes hiperparâmetros:

| Hiperparâmetro | Espaço de busca | Função |
|---|---|---|
| `learning_rate` | 0,01 a 0,3, escala logarítmica | Controla a contribuição de cada nova árvore para o modelo. |
| `max_iter` | 100 a 500, passo 50 | Determina o número máximo de iterações de boosting. |
| `max_leaf_nodes` | Inteiros de 15 a 63 | Limita o número máximo de folhas por árvore. |
| `max_depth` | `None`, 4, 6, 8 ou 12 | Limita a profundidade máxima das árvores. |
| `min_samples_leaf` | Inteiros de 10 a 60 | Determina a quantidade mínima de amostras por folha. |
| `l2_regularization` | \(10^{-5}\) a 10, escala logarítmica | Controla a intensidade da regularização L2 aplicada aos valores das folhas. |

O hiperparâmetro `learning_rate` controla a intensidade das correções realizadas por cada nova árvore. Valores menores geralmente exigem mais iterações para produzir correções acumuladas comparáveis.

Por esse motivo, sua investigação está relacionada à escolha de `max_iter`, que estabelece a quantidade de etapas de boosting.

Os hiperparâmetros `max_leaf_nodes`, `max_depth` e `min_samples_leaf` controlam a complexidade das árvores utilizadas no processo. Já `l2_regularization` introduz uma penalização destinada a restringir os valores previstos pelas folhas, contribuindo para o controle da complexidade do modelo.

O parâmetro `early_stopping` foi fixado como `False`, de forma que a quantidade de iterações seja determinada pelo hiperparâmetro `max_iter`, sem a criação automática de uma partição interna de validação que não siga o agrupamento químico adotado no projeto.

#### <a id='toc2_5_2_8_'></a>[XGBoost](#toc0_)

O *Extreme Gradient Boosting* (XGBoost) também utiliza a construção sequencial de árvores, incorporando mecanismos de regularização e estratégias computacionais para melhorar a eficiência do treinamento.

No presente trabalho, foi utilizada a implementação `XGBRegressor`, com a função objetivo `reg:squarederror` e o método de construção de árvores `hist`.

Foram otimizados os seguintes hiperparâmetros:

| Hiperparâmetro | Espaço de busca | Função |
|---|---|---|
| `n_estimators` | 100 a 600, passo 50 | Determina a quantidade de etapas de boosting. |
| `learning_rate` | 0,01 a 0,3, escala logarítmica | Controla a contribuição de cada nova árvore. |
| `max_depth` | Inteiros de 3 a 10 | Limita a profundidade das árvores. |
| `min_child_weight` | 1 a 15, escala logarítmica | Estabelece o peso mínimo acumulado associado a um nó filho, com base na informação de segunda ordem da função de perda. |
| `subsample` | 0,6 a 1,0 | Define a fração de registros utilizada em cada etapa de boosting. |
| `colsample_bytree` | 0,6 a 1,0 | Define a fração de atributos amostrada para cada árvore. |
| `reg_alpha` | \(10^{-5}\) a 10, escala logarítmica | Controla a regularização L1. |
| `reg_lambda` | \(10^{-3}\) a 100, escala logarítmica | Controla a regularização L2. |

Os hiperparâmetros `n_estimators` e `learning_rate` atuam conjuntamente sobre o processo de aprendizagem. Em geral, taxas de aprendizagem menores podem exigir mais árvores, enquanto taxas mais elevadas produzem correções mais intensas a cada etapa.

O parâmetro `max_depth` controla a complexidade das árvores. Já `min_child_weight` restringe a criação de determinadas divisões com base na soma dos valores de segunda ordem associados às observações, contribuindo para regular o crescimento das árvores.

Os parâmetros `subsample` e `colsample_bytree` introduzem amostragem de registros e atributos, respectivamente, enquanto `reg_alpha` e `reg_lambda` controlam diferentes formas de regularização aplicadas aos pesos das folhas.

A otimização conjunta desses parâmetros permite investigar diferentes compromissos entre capacidade de ajuste, complexidade e generalização.

### <a id='toc2_6_2_'></a>[Evolução da otimização](#toc0_)

Vamos utilizar os CSVs para construir gráficos com duas informações:

- Menor MAE acumulado até aquele trial.

Temos o código dos gráficos abaixo, faremos um gráfico conjunto para observarmos as diferentes melhoras de cada modelo.

```python
PASTA_TRIALS = Path("tiamat_hpc/results/trials")

arquivos = {
    "Random Forest": "random_forest.csv",
    "Extra Trees": "extra_trees.csv",
    "HistGradientBoosting": "hist_gradient_boosting.csv",
    "XGBoost": "xgboost_parcial.csv",
    "Ridge Regression": "ridge.csv"
}

cores = {
    "Random Forest": "royalblue",
    "Extra Trees": "forestgreen",
    "HistGradientBoosting": "darkorange",
    "XGBoost": "firebrick",
    "Ridge Regression": "purple"
}

fig, ax = plt.subplots(figsize=(11, 6))

for modelo, nome_arquivo in arquivos.items():

    caminho = PASTA_TRIALS / nome_arquivo

    df = pd.read_csv(caminho)

   
    df = df.loc[
        (df["state"].astype(str).str.upper() == "COMPLETE")
        & df["value"].notna()
    ].copy()

    if df.empty:
        print(f"Nenhum trial válido: {modelo}")
        continue


    df = df.sort_values("number")

    df["melhor_mae"] = df["value"].cummin()

    ax.step(
        df["number"],
        df["melhor_mae"],
        where="post",
        label=modelo,
        color=cores[modelo],
        linewidth=2
    )

    melhor = df.loc[df["value"].idxmin()]

    ax.scatter(
        melhor["number"],
        melhor["value"],
        color=cores[modelo],
        marker="o",
        s=55,
        zorder=5
    )

ax.set_title(
    "Evolução da otimização dos hiperparâmetros",
    fontsize=14
)

ax.set_xlabel("Número do trial", fontsize=12)
ax.set_ylabel("Melhor MAE acumulado (K)", fontsize=12)

ax.grid(True, alpha=0.25)
ax.legend(title="Modelo")

plt.tight_layout()
plt.show()
```

### <a id='toc2_6_3_'></a>[Influência da seleção de atributos](#toc0_)

Até agora, identificamos um resultado interessante: os melhores trials dos cinco modelos utilizaram todos os 81 atributos. Entretanto, precisamos investigar se isso representa uma tendência geral ou apenas uma característica das melhores configurações encontradas. Nesse sentido, podemos perguntar: a utilização do SelectKBest esteve associada a uma redução do MAE durante a otimização dos hiperparâmetros?

Para investigar isso, compararemos os trials que utilizaram seleção de atributos com aqueles que conservaram todos os atributos.

É importante reconhecer uma limitação: o Optuna modificou simultaneamente vários hiperparâmetros. Portanto, essa comparação será observacional e não permitirá atribuir diferenças de desempenho exclusivamente à seleção de atributos.

Abaixo temos o código para comparar os grupos. Utilizaremos os CSVs que já estamos trabalhando no notebook.

```python
resultados_selecao = []

for modelo, nome_arquivo in arquivos.items():

    df = pd.read_csv(PASTA_TRIALS / nome_arquivo)

    df = df.loc[
        (df["state"].astype(str).str.upper() == "COMPLETE")
        & df["value"].notna()
    ].copy()

    if df.empty:
        continue

    df["selecao"] = df["params_usar_selecao"].map({
        True: "Com seleção",
        False: "Sem seleção",
        "True": "Com seleção",
        "False": "Sem seleção"
    })

    resumo = (
        df.groupby("selecao")["value"]
        .agg(["count", "mean", "median", "std", "min"])
        .reset_index()
    )

    resumo["Modelo"] = modelo
    resultados_selecao.append(resumo)

df_selecao = pd.concat(
    resultados_selecao,
    ignore_index=True
)

display(
    df_selecao[
        ["Modelo", "selecao", "count", "mean",
         "median", "std", "min"]
    ].round(4)
)
```

## <a id='toc2_7_'></a>[Indução dos modelos otimizados](#toc0_)

Após a etapa de otimização, os cinco algoritmos de regressão foram considerados para a avaliação final: Ridge Regression, Random Forest, Extra Trees, HistGradientBoosting e XGBoost.

Durante a otimização, cada configuração de hiperparâmetros foi avaliada por validação cruzada agrupada com cinco folds, utilizando o MAE médio como critério de seleção. Para cada algoritmo, foi identificada a configuração que apresentou o menor MAE médio entre os trials concluídos. Essas configurações serão utilizadas para reconstruir os respectivos pipelines, preservando tanto os hiperparâmetros dos estimadores quanto as decisões referentes à seleção de atributos.

A reconstrução dos pipelines não envolve, nesta etapa, o ajuste dos modelos ou a utilização do conjunto de teste. O treinamento final será realizado posteriormente com todo o conjunto de treinamento, mantendo os dados de teste reservados para a avaliação da capacidade de generalização.

Esse procedimento permite separar a seleção de hiperparâmetros da avaliação final, reduzindo o risco de utilizar informações do conjunto de teste durante o desenvolvimento dos modelos.

Inicialmente, vamos recuperar os melhores hiperparâmetros para cada modelo.

```python
PASTA_TRIALS = Path("tiamat_hpc/results/trials")

arquivos = {
    "Random Forest": "random_forest.csv",
    "Extra Trees": "extra_trees.csv",
    "HistGradientBoosting": "hist_gradient_boosting.csv",
    "XGBoost": "xgboost_parcial.csv",
    "Ridge": "ridge.csv"
}

melhores_parametros = {}

for modelo, arquivo in arquivos.items():

    df = pd.read_csv(PASTA_TRIALS / arquivo)

    df = df.loc[
        (df["state"].astype(str).str.upper() == "COMPLETE")
        & df["value"].notna()
    ].copy()

    if df.empty:
        raise ValueError(f"Nenhum trial completo para {modelo}")

    melhor = df.loc[df["value"].idxmin()]

    parametros = {}

    for coluna in df.columns:
        if coluna.startswith("params_"):
            valor = melhor[coluna]

            if pd.notna(valor):
                parametros[coluna.removeprefix("params_")] = valor

    melhores_parametros[modelo] = parametros

    print(f"\n{modelo}")
    print("Melhor trial:", int(melhor["number"]))
    print("MAE (CV):", round(float(melhor["value"]), 4))
    print("Parâmetros:", parametros)
```

### <a id='toc2_7_1_'></a>[Reconstrução e verificação dos pipelines](#toc0_)

Como todos os melhores trials utilizaram os 81 atributos, manteremos k="all".

```python
from sklearn.base import clone

modelos_otimizados = {}

for nome, parametros in melhores_parametros.items():

    pipeline = construir_pipeline_com_selecao(
        pipelines[nome],
        k="all"
    )

    parametros_modelo = {
        f"modelo__{chave}": valor
        for chave, valor in parametros.items()
        if chave not in {"usar_selecao", "k"}
    }

    pipeline.set_params(**parametros_modelo)

    modelos_otimizados[nome] = pipeline
```

```python
def calcular_metricas(y_real, y_previsto):

    y_real = np.asarray(y_real, dtype=float).ravel()
    y_previsto = np.asarray(y_previsto, dtype=float).ravel()

    if y_real.shape != y_previsto.shape:
        raise ValueError("As dimensões devem ser iguais.")

    
    mae = mean_absolute_error(y_real, y_previsto)

    
    rmse = np.sqrt(mean_squared_error(y_real, y_previsto))

    maape = calcular_maape(y_real, y_previsto)

    return {
        "MAE (K)": mae,
        "RMSE (K)": rmse,
        "MAAPE (rad)": maape
    }
```

## <a id='toc2_8_'></a>[Interpretabilidade e discussão científica](#toc0_)

### <a id='toc2_8_1_'></a>[Fundamentos da interpretabilidade: Como funciona o Permutation Importance?](#toc0_)

O princípio é simples: se determinado atributo contém informação importante para as previsões de um modelo, embaralhar seus valores deve prejudicar o desempenho.

O procedimento será:
1. Calcular o MAE do modelo treinado sobre os dados de avaliação.
2. Embaralhar os valores de um atributo, mantendo os demais intactos.
3. Calcular novamente o MAE.
4. Medir quanto o erro aumentou.
5. Repetir a permutação diversas vezes e calcular a importância média.

Matematicamente, teremos a seguinte fórmula da importância:

$$
I_j = \frac{1}{R}\sum_{r=1}^{R}
\left(
\mathrm{MAE}_{j,r}^{\mathrm{permutado}}
-
\mathrm{MAE}_{\mathrm{original}}
\right)
$$

Aqui, $I_j$ representa a importância do atributo $j$, e $R$ é o número de repetições.

Se a importância for positiva e elevada, o embaralhamento prejudicou significativamente as previsões. Se for próxima de zero, o modelo sofreu pouca alteração. Valores negativos indicam que o embaralhamento melhorou o desempenho naquela avaliação, algo que também pode ocorrer.

Para o leitor mais curioso, leia sobre [Permutation importance](https://scikit-learn.org/stable/modules/permutation_importance) no Scikit-learn.


### <a id='toc2_8_2_'></a>[Primeira implementação - Extra Trees](#toc0_)

Como o notebook já possui modelos_treinados, X_teste e y_teste, não precisamos reconstruir nem treinar novamente o Extra Trees.

```python
from sklearn.inspection import permutation_importance

resultado_pi = permutation_importance(
    estimator=modelos_treinados["Extra Trees"],
    X=X_teste,
    y=y_teste,
    scoring="neg_mean_absolute_error",
    n_repeats=10,
    random_state=SEED,
    n_jobs=-1
)
```

As dez repetições permitem observar a variabilidade da importância associada aos diferentes embaralhamentos. Essa variabilidade não deve ser confundida com um intervalo de confiança estatístico. O cálculo utiliza o modelo já treinado e não exige novos ajustes. Como estamos usando o conjunto de teste para a interpretação, seus resultados deverão ser tratados como uma análise descritiva do modelo final, sem utilizá-los para realizar novas escolhas de hiperparâmetros ou atributos. 

Na célula abaixo, iremos organizar os resultados obtidos.

```python
df_importancias = pd.DataFrame({
    "Atributo": X_teste.columns,
    "Importancia_media_K": resultado_pi.importances_mean,
    "Desvio_padrao_K": resultado_pi.importances_std
})

df_importancias = (
    df_importancias
    .sort_values("Importancia_media_K", ascending=False)
    .reset_index(drop=True)
)

display(df_importancias.head(15))
```

Essa tabela mostra os 15 descritores cujo embaralhamento provocou, em média, os maiores aumentos no MAE. 

No Capítulo 2, identificamos correlações de Pearson próximas de 1 entre alguns pares de atributos. Isso cria uma limitação importante: se dois atributos carregam informações semelhantes, o modelo pode continuar prevendo relativamente bem quando apenas um deles é embaralhado. Assim, ambos podem apresentar importâncias individuais pequenas, mesmo sendo preditivamente relevantes em conjunto.

Por isso, antes de atribuir um significado físico aos descritores mais importantes, investigaremos também sua redundância e sua relação com as propriedades elementares que deram origem aos 81 atributos. __Não interpretaremos uma importância elevada como prova de causalidade.__

### <a id='toc2_8_3_'></a>[Segunda implementação - XGBoost](#toc0_)

Agora, repetiremos o mesmo processo para o XGBoost, visando complementar os resultados.

```python
resultado_pi_xgb = permutation_importance(
    estimator=modelos_treinados["XGBoost"],
    X=X_teste,
    y=y_teste,
    scoring="neg_mean_absolute_error",
    n_repeats=10,
    random_state=SEED,
    n_jobs=-1
)
```

Organizando os resultados:

```python
df_importancias_xgb = pd.DataFrame({
    "Atributo": X_teste.columns,
    "Importancia_media_K": resultado_pi_xgb.importances_mean,
    "Desvio_padrao_K": resultado_pi_xgb.importances_std
})

df_importancias_xgb = df_importancias_xgb.sort_values(
    "Importancia_media_K",
    ascending=False
).reset_index(drop=True)

display(df_importancias_xgb.head(15))
```

Além disso, a existência de correlações entre os atributos pode influenciar os valores de importância obtidos, uma vez que diferentes variáveis podem conter informações parcialmente redundantes. ### <a id='toc2_8_5_'></a>[Discussão físico-química dos descritores](#toc0_)

Depois dessa comparação, passaremos à interpretação das propriedades que deram origem aos atributos destacados. O dataset contém descritores associados a propriedades elementares, como massa atômica, energia de ionização, raio atômico, densidade, afinidade eletrônica e condutividade térmica, entre outras. Logo, o objetivo será compreender por que essas informações podem ajudar a distinguir materiais com diferentes temperaturas críticas.

Por exemplo:

| Atributo | O que representa |
|-----------|------------------|
| mean_atomic_mas | Média das massas atômicas dos elementos presentes no material. |
| wtd_mean_atomic_mass | Média das massas atômicas ponderada pelas proporções de cada elemento na composição. |
| entropy_fie | Entropia estatística associada à distribuição das primeiras energias de ionização dos elementos. |
| range_atomic_radius | Diferença entre o maior e o menor raio atômico dos elementos presentes. |
| wtd_std_ThermalConductivity | Desvio-padrão ponderado das condutividades térmicas dos elementos que compõem o material. |

É fundamental observar que entropy_fie, por exemplo, não representa diretamente a entropia termodinâmica do supercondutor. É um descritor estatístico calculado a partir de uma propriedade elementar. Da mesma forma, wtd_mean_ThermalConductivity não corresponde necessariamente à condutividade térmica medida do material composto.

Essa distinção evita interpretações físicas equivocadas.

### <a id='toc2_8_8_'></a>[Interpretação físico-química dos atributos relacionados à condutividade térmica](#toc0_)

A análise da importância por permutação revelou que os descritores relacionados à condutividade térmica dos elementos químicos apresentam elevada relevância preditiva para os modelos Extra Trees e XGBoost. Entre esses atributos, destacou-se range_ThermalConductivity, cuja permutação provocou um aumento médio do MAE de 3,737 K no Extra Trees e 9,232 K no XGBoost, representando a maior importância individual observada nos dois algoritmos.

Esse atributo corresponde à amplitude dos valores tabulados de condutividade térmica dos elementos constituintes de cada composição química, sendo definido pela diferença entre os valores máximo e mínimo:

$$
\operatorname{range}(\kappa)=\kappa_{\max}-\kappa_{\min}
$$

em que $\kappa_{\max}$ e $\kappa_{\min}$ representam, respectivamente, os maiores e menores valores de condutividade térmica entre os elementos presentes na composição. Portanto, o atributo descreve a heterogeneidade das propriedades térmicas elementares, e não a condutividade térmica efetiva do material supercondutor.

Além desse descritor, outras variáveis relacionadas à condutividade térmica apresentaram contribuições relevantes. O desvio-padrão ponderado (wtd_std_ThermalConductivity) provocou aumentos médios do MAE de 2,579 K no Extra Trees e 1,952 K no XGBoost. A média ponderada (wtd_mean_ThermalConductivity) apresentou aumentos de 0,687 K e 0,551 K, respectivamente, enquanto a entropia ponderada (wtd_entropy_ThermalConductivity) resultou em aumentos de 0,637 K e 0,769 K.

Esses resultados indicam que diferentes características da distribuição das condutividades térmicas elementares fornecem informações utilizadas pelos modelos para estimar a temperatura crítica. A maior importância de range_ThermalConductivity, especialmente no XGBoost, sugere uma dependência preditiva acentuada em relação a esse descritor. Entretanto, a importância por permutação não permite determinar a direção dessa associação, de modo que não é possível afirmar que maiores diferenças entre as condutividades térmicas dos elementos estejam associadas a aumentos ou reduções da temperatura crítica.

Do ponto de vista físico-químico, uma hipótese plausível é que a heterogeneidade das propriedades térmicas elementares esteja associada a características das composições químicas e das diferentes famílias de supercondutores presentes no conjunto de dados. Nesse contexto, os descritores podem funcionar como indicadores indiretos de propriedades estruturais, eletrônicas ou vibracionais que não estão explicitamente representadas nas variáveis utilizadas pelos modelos. Contudo, essa interpretação permanece hipotética, uma vez que os resultados obtidos não permitem identificar os mecanismos físicos responsáveis pelas associações observadas.

É importante ressaltar que a condutividade térmica efetiva de um material e sua temperatura crítica são propriedades físicas distintas. Embora o transporte de calor possa influenciar a estabilidade térmica de dispositivos supercondutores em determinadas condições de operação, isso não demonstra que a condutividade térmica determine diretamente a temperatura de transição para o estado supercondutor. Além disso, os atributos analisados são calculados a partir de propriedades tabuladas dos elementos químicos, não correspondendo necessariamente às propriedades experimentais dos compostos formados.

### <a id='toc2_8_9_'></a>[Raio atômico e massa atômica](#toc0_)

Primordialmente, range_atomic_radius representa a diferença entre o maior e o menor raio atômico dos elementos presentes em uma composição:

$$
\Delta r=r_{\max}-r_{\min}
$$

std_atomic_mass representa o desvio-padrão das massas atômicas elementares. No dataset, esse descritor não é ponderado pela proporção estequiométrica dos elementos.

Os dois atributos descrevem formas de heterogeneidade química, mas não são equivalentes: um mede a amplitude dos tamanhos atômicos e o outro mede a dispersão das massas.

#### <a id='toc2_8_9_1_'></a>[Relação com supercondutividade](#toc0_)

O raio atômico pode estar associado a diferenças de tamanho entre os elementos que constituem um material. Essas diferenças podem ser relevantes para aspectos estruturais, como distorções locais e possibilidades de organização cristalina. Entretanto, não podemos deduzir a estrutura cristalina a partir da amplitude dos raios atômicos. Dois materiais com composições semelhantes podem possuir estruturas distintas e temperaturas críticas diferentes.

A massa atômica, por sua vez, tem uma conexão conceitual importante com a física da supercondutividade convencional: a massa dos íons influencia as frequências das vibrações da rede cristalina, os chamados fônons. Na teoria BCS, a interação entre elétrons mediada por fônons desempenha papel fundamental na formação dos pares de Cooper. Contudo, os efeitos da massa sobre $T_c$ dependem das propriedades do material e do mecanismo de supercondutividade.

O resultado encontrado não demonstra que a dispersão das massas atômicas controle diretamente $T_c$. Ele indica que esse descritor contém informação preditiva utilizada pelos modelos.

#### <a id='toc2_8_9_2_'></a>[Respaldo na literatura](#toc0_)

Há um resultado interessante no artigo original de [Kam Hamidieh (2018)](https://doi.org/10.1016/j.commatsci.2018.07.052): propriedades relacionadas à condutividade térmica, ao raio atômico, à valência, à afinidade eletrônica e à massa atômica também contribuíram para a capacidade preditiva do modelo investigado pelo autor.

### <a id='toc2_8_10_'></a>[Valência e primeira energia de ionização](#toc0_)

Há uma diferença interessante: o Extra Trees apresentou maior importância para a amplitude da primeira energia de ionização, enquanto o XGBoost atribuiu maior importância às médias ponderadas de valência. Isso sugere que os dois algoritmos podem utilizar aspectos distintos das informações relacionadas às propriedades eletrônicas dos elementos.

Mas ainda não sabemos se essas diferenças decorrem de relações fisicamente significativas, da redundância entre descritores ou da maneira como cada algoritmo constrói suas previsões.

#### <a id='toc2_8_10_1_'></a>[Primeira energia de ionização: range fie](#toc0_)

O atributo range_fie representa a diferença entre o maior e o menor valor dessa propriedade entre os elementos constituintes:

$$
\Delta E_{\mathrm{ion}} =
E_{\mathrm{ion,max}}-E_{\mathrm{ion,min}}
$$

Uma amplitude elevada significa que a composição reúne elementos com diferentes energias de ionização. Do ponto de vista físico-químico, essas diferenças podem estar associadas ao caráter químico dos elementos e, indiretamente, à natureza das ligações que podem estabelecer.

Entretanto, a energia de ionização de um átomo isolado não corresponde diretamente à energia necessária para remover um elétron de um sólido. Nos sólidos, os estados eletrônicos dependem de fatores como estrutura cristalina, hibridização orbital e formação de bandas de energia. Por isso, range_fie pode funcionar como um descritor indireto da composição química, mas não permite determinar, isoladamente, a estrutura eletrônica do supercondutor.

#### <a id='toc2_8_10_2_'></a>[Valência: wtd_mean_Valence, wtd_gmean_Valence e wtd_std_Valence](#toc0_)

No dataset utilizado pela Tiamat, a valência é uma propriedade elementar empregada para construir os descritores estatísticos da composição química. Não devemos confundi-la automaticamente com o estado de oxidação de cada elemento no composto. A importância conjunta desses três descritores sugere que tanto os valores típicos quanto a heterogeneidade das valências elementares contêm informações úteis para prever a temperatura crítica.

Isso é compatível com a ideia geral de que a composição química está associada às propriedades eletrônicas dos materiais. Contudo, é necessário fazer uma distinção física importante.

>Valência elementar não é o mesmo que concentração de portadores de carga em um supercondutor.

Em diversas famílias de supercondutores, especialmente nos cupratos, a concentração de portadores e a dopagem desempenham papéis fundamentais no comportamento de $T_c$. Entretanto, não podemos obter diretamente a concentração de portadores apenas calculando a média das valências elementares. São necessárias informações adicionais sobre os estados de oxidação, a estrutura eletrônica e o material estudado. Por isso, uma hipótese plausível é que os descritores de valência ajudem os algoritmos a distinguir diferentes composições e famílias de supercondutores. Os resultados não demonstram que aumentar a valência média necessariamente aumentaria $T_c$.

#### <a id='toc2_8_10_3_'></a>[Respaldo na literatura](#toc0_)

Em 1955, Bernd T. Matthias publicou o artigo [ Empirical Relation between Superconductivity and the Number of Valence Electrons per Atom](https://doi.org/10.1103/PhysRev.97.74).

O trabalho investigou experimentalmente a relação entre a temperatura crítica e o número médio de elétrons de valência por átomo, identificando regiões favoráveis à supercondutividade próximas de cinco e sete elétrons de valência por átomo nos sistemas analisados. 

Essa observação ficou conhecida como parte das regras empíricas de Matthias.

É uma referência particularmente útil, porque demonstra que o interesse pelo número de elétrons de valência como descritor da supercondutividade antecede o aprendizado de máquina em muitas décadas.

## <a id='toc2_10_'></a>[Conclusões e principais aprendizados.](#toc0_)

Os resultados demonstraram a viabilidade da utilização de algoritmos de aprendizado de máquina para prever a temperatura crítica de supercondutores a partir de descritores de suas composições químicas. Entre os cinco modelos avaliados, o Extra Trees apresentou o melhor desempenho no conjunto de teste, com MAE de 5,411 K, seguido pelo XGBoost, com 5,508 K. Os modelos baseados em árvores apresentaram erros consideravelmente inferiores aos do Ridge Regression, sugerindo a relevância de relações não lineares entre os atributos e a temperatura crítica. Além disso, a manutenção dos 81 atributos mostrou-se vantajosa nas configurações investigadas, enquanto a análise de importância por permutação destacou descritores relacionados à condutividade térmica, ao raio atômico e à valência. Entretanto, a relevância preditiva dessas propriedades não estabelece relações causais com os mecanismos físicos da supercondutividade. Conclui-se, portanto, que o aprendizado de máquina constitui uma ferramenta promissora para a estimativa da temperatura crítica, embora a ausência de informações estruturais e eletrônicas dos materiais limite a interpretação física dos resultados e sua generalização para composições distintas das investigadas.

A partir das hipóteses iniciais:

1. Modelos baseados em árvores e ensembles apresentam melhor capacidade preditiva do que regressões lineares, devido às relações complexas entre composição e propriedades supercondutoras.

2. Um subconjunto selecionado dos 81 descritores pode alcançar desempenho próximo ao conjunto completo, reduzindo complexidade e facilitando interpretação.

3. Determinados grupos de descritores, como estatísticas de elétrons de valência ou propriedades atômicas, apresentam maior contribuição preditiva do que outros.

Temos que a primeira hipótese tem fortes indícios a partir dos resultados encontrados, de forma que eles corroboram a suposição. A segunda hipótese também é sustentada pelos resultados encontrados, visto que, apesar do melhor desempenho final de todos os algoritmos sem a seleção de atributos, um subconjunto pôde se aproximar do desempenho do conjunto completo. Por fim, a hipótese 3 também é parcialmente confirmada pelos resultados, porque alguns conjuntos descritores realmente apresentam maior contribuição para a predição.

Em se tratando dos aprendizados metodológicos e experimentais, a utilização do HPC da Ilum foi um dos principais. Isso porque, desde a aula introdutória de como utilizar o ambiente, não foi possível praticar a utilização do SLURM, por isso a prática durante a produção desse trabalho foi essencial para compreender melhor essa habilidade. Além disso, também é possível citar que esse trabalho fomentou duas das principais habilidades de um cientista: a leitura de artigos buscando informações pertinentes e a capacidade de guiar o leitor em uma linha de raciocínio clara. Dessa forma, demais aprendizados que não foram citados, como a sedimentação dos conhecimentos da matéria e a compreensão (mesmo que superficial) do mecanismo físico dos supercondutores, também foram adquiridos durante a produção do Tiamat.

Por fim, algumas limitações metodológicas devem ser consideradas. Os modelos utilizaram exclusivamente descritores derivados das composições químicas, sem incorporar informações sobre estrutura cristalina, propriedades eletrônicas ou condições experimentais, fatores que podem influenciar significativamente a temperatura crítica. Embora os resultados no conjunto de teste indiquem uma capacidade de generalização satisfatória, não garantem desempenho equivalente para novas famílias de supercondutores.
