### Supercondutividade e Temperatura crítica (TC)

Nesse notebook, trabalharemos com um problema científico que deve ser compreendido pelo leitor para que possamos seguir com a utilização de Machine Learning. Como pode ser observado pelo título, faremos a predição da temperatura crítica de supercondutores, mas a pergunta inicial é: o que são os objetos que serão estudados?

A princípio, a supercondutividade é um fenômeno físico em que certos materiais perdem completamente a resistência elétrica e expulsam campos magnéticos de seu interior quando resfriados abaixo de uma temperatura crítica específica. Esses materiais apresentam algumas características interessantes:
<br><br>

• __Resistência elétrica nula__: A corrente elétrica pode fluir por um circuito fechado de material supercondutor indefinidamente, sem perda de energia ou dissipação de calor por efeito Joule;


• __Efeito Meissner__: Os supercondutores atuam como diamagnéticos perfeitos, expulsando campos magnéticos e permitindo fenômenos como a levitação magnética.
<br><br>


Enquanto isso, a temperatura crítica, é a temperatura limite abaixo da qual um material transiciona para esse estado de supercondutividade, perdendo toda a sua resistência elétrica. Acima dessa temperatura, o material se comporta como um condutor comum (gerando calor e oferecendo resistência à corrente elétrica). Abaixo dela, ocorre uma mudança de fase termodinâmica abrupta que ativa as propriedades supercondutoras.

A temperatura crítica também apresenta características que devem ser citadas:
<br><br>

• __Ela é específica para cada material__: Cada elemento ou composto possui a sua própria \(T_{c}\). Por exemplo, o mercúrio puro se torna supercondutor a 4,2 K (-268,95 °C), enquanto cerâmicas complexas (conhecidos como supercondutores de alta temperatura) podem atingir esse estado acima de 130 K (-143 °C).


• __Dependência magnética__: A temperatura crítica não é fixa se o material estiver exposto a um campo magnético. Se o campo magnético externo for muito forte (chamado de campo crítico), ele pode destruir o estado supercondutor mesmo que o material esteja abaixo da \(T_{c}\).


• __O "Santo Graal" da Física__: Atualmente, os supercondutores práticos exigem refrigeração cara com hélio ou nitrogênio líquido. Encontrar um material com uma temperatura crítica que opere na temperatura ambiente (cerca de 293 K ou 20 °C) revolucionaria a tecnologia global, permitindo redes elétricas sem nenhuma perda de energia.
<br><br>


Com isso, os supercondutores possuem aplicações revolucionárias devido às suas propriedades. Na medicina, eles alimentam os potentes ímãs dos aparelhos de ressonância magnética, permitindo imagens diagnósticas de alta resolução. No transporte, viabilizam os trens de levitação magnética (Maglev), que flutuam sobre os trilhos e viajam a velocidades altíssimas sem atrito. No setor de energia, abrem caminho para redes de transmissão elétrica sem perdas de carga e são fundamentais nos reatores de fusão nuclear Tokamak, que usam campos magnéticos para confinar o plasma superaquecido. Por fim, na computação, são a base para o desenvolvimento de computadores quânticos e processadores ultrarrápidos.


### Fatores físico-químicos relacionados à supercondutividade

Existe um estrito conjunto de fatores físico-químicos que determinam se um material pode entrar no estado quântico macroscópico de supercondutividade, evidenciando que a interação entre a estrutura da matéria, a temperatura e as forças externas molda o comportamento do material. 

Fundamentalmente, existem três fatores físicos críticos que determinam se o material retornará (ou não) ao seu estado condutor normal:
<br><br>

• __Temperatura Crítica ($T_{c}$)__: É a temperatura abaixo da qual a agitação térmica é baixa o suficiente para permitir a atração indireta entre elétrons. Varia de quase zero absoluto (como o Tungstênio a 0,015 K) até patamares mais altos em cerâmicas complexas (cupratos);

• __Campo Magnético Crítico ($H_{c}$)__: Campos magnéticos externos destroem o estado supercondutor. Supercondutores do Tipo I possuem um único limite rígido ($H_{c}$). Já os do Tipo II toleram um estado misto (vórtices quantizados) entre um campo inferior ($H_{c1}$) e superior ($H_{c2}$), tornando-os úteis para gerar campos magnéticos massivos.

• __Densidade de Corrente Crítica ($J_{c}$)__: A quantidade máxima de corrente elétrica por unidade de área que o material transporta sem resistência. Correntes que excedem esse limite geram campos magnéticos induzidos que destroem a supercondutividade.

<br><br>


Além disso, a composição química e a organização atômica influenciam diretamente os mecanismos quânticos que estabilizam o fenômeno:

<br><br>

• __Aceleração Fônon-Elétron (Rede Cristalina)__: Na teoria convencional (BCS), a passagem de um elétron deforma localmente os íons positivos da rede cristalina. Essa distorção (vibração molecular ou fônon) cria uma região temporária de carga positiva que atrai um segundo elétron, superando a repulsão natural de Coulomb. Esse par ligado chama-se __Par de Cooper__;


• __Anisotropia e Planos de Óxido de Cobre ($\text{CuO}_{2}$)__: Em supercondutores de alta temperatura (cupratos), a condução ocorre de forma bidimensional ao longo de planos de óxido de cobre intercalados por camadas isolantes. A geometria e a distância entre esses planos ditam a eficácia da supercondutividade;


• __Dopagem Química__: Pequenas alterações estequiométricas mudam drasticamente o comportamento do material. A introdução controlada de átomos (como oxigênio em cerâmicas ou dopantes químicos em fulerenos) ajusta a densidade de portadores de carga até atingir um ponto ideal de transição eletrônica.


• __Impurezas Magnéticas__: Elementos com propriedades magnéticas fortes (como manganês ou ferro) quebram a simetria de spin dos Pares de Cooper, atuando como venenos químicos que destroem ou reduzem severamente a $T_{c}$ do material.


<br><br>

Por fim, a aplicação de __pressões mecânicas extremas__ também pode alterar a supercondutividade. Ao comprimir a rede cristalina, as frequências de vibração dos átomos (fônons) são alteradas.

### Por que utilizar Machine Learning?

Apesar da existência de teorias importantes como a __BCS__, elas ainda falham em explicar e prever as propriedades de alguns tipos de supercondutores, como, por exemplo, os supercondutores de alta temperatura.

A partir dessas propriedades, foram criados descritores estatísticos: médias, médias ponderadas, médias geométricas, entropias, amplitudes, desvios padrão, entre outros... No artigo, o autor dá um exemplo de como as features foram extraídas do banco de dados.

Considerando o supercondutor $\text{Re}_{7}{Zr}_{1}$ com $T_c$ = 6.7 K :

A condutividade térmica do Rênio e do zircônia são $t_1$ = 48 e $t_2$ = 23 $\frac{W}{(m\times K)}$, respectivamente. 

Dessa forma, as razões entre os elementos no material são utilizadas para definir características:

$$
p_1 = \frac{6}{6+1} = \frac{6}{7},
\qquad
p_2 = \frac{1}{6+1} = \frac{1}{7}.

$$

As frações das condutividades térmicas totais também são utilizadas:

$$
w_1 = \frac{t_1}{t_1+t_2}
= \frac{48}{48+23}
= \frac{48}{71},
$$

$$
w_2 = \frac{t_2}{t_1+t_2}
= \frac{23}{48+23}
= \frac{23}{71}.
$$

Precisamos de alguns valores intermediários baseados nas Eqs.~(1) e (2):

$$
A =
\frac{p_1w_1}{p_1w_1+p_2w_2}
\approx 0.926,
$$

$$
B =
\frac{p_2w_2}{p_1w_1+p_2w_2}
\approx 0.074.
$$

Após obtermos os valores de $p_1$, $p_2$, $w_1$, $w_2$, $A$ e $B$, torna-se possível extrair 10 features a partir das condutividades térmicas de Rênio e Zircônio, conforme apresentado na Tabela 2.

<br><br>

#### Correlação dos atributos com a temperatura crítica

Depois de compreender a variável-alvo, nosso próximo objetivo será identificar quais descritores possuem associações lineares mais fortes com $T_c$.
Utilizaremos a correlação de Pearson:

$$
r_{X,Y}=
\frac{\operatorname{Cov}(X,Y)}
{\sigma_X\sigma_Y}
$$

Seu valor está entre $-1$ e $1$. Valores próximos aos extremos indicam associações lineares fortes, enquanto valores próximos de zero indicam pouca associação linear, mas não necessariamente ausência de relação.

Para a plotagem, selecionaremos os 15 maiores valores de correlação em módulo.

Esse gráfico será exploratório, não um procedimento de seleção de atributos para treinamento. Se decidirmos usar correlação para selecionar preditores, a seleção deverá ocorrer exclusivamente nos dados de treinamento, respeitando a validação cruzada.

````python

correlacoes = (
    df.corr(numeric_only=True)["critical_temp"]
    .drop("critical_temp")
)


top_atributos = (
    correlacoes.abs()
    .sort_values(ascending=False)
    .head(15)
    .index
)

top_correlacoes = correlacoes.loc[top_atributos].sort_values()

fig, ax = plt.subplots(figsize=(10, 6))

ax.barh(
    top_correlacoes.index,
    top_correlacoes.values,
    color="steelblue"
)

ax.axvline(0, color="black", linewidth=0.8)
ax.set_title("15 atributos mais correlacionados com Tc")
ax.set_xlabel("Coeficiente de correlação de Pearson")
ax.set_ylabel("Atributos físico-químicos")
ax.set_xlim(-1, 1)
ax.grid(axis="x", alpha=0.2)

plt.tight_layout()
plt.show()
````

A partir desses atributos, podemos notar inicialmente que existem variáveis com correlações relativamente fortes com a temperatura crítica ($T_c$​). Entretanto, vale reforçar que essas observações têm caráter exploratório e não serão utilizadas, neste momento, para seleção de atributos ou construção do modelo.

Observa-se que as maiores correlações positivas estão associadas a medidas de dispersão e heterogeneidade dos elementos presentes no material, como desvios-padrão, amplitudes (ranges) e entropias de propriedades físico-químicas. Isso sugere que materiais cujos constituintes apresentam maior diversidade em determinadas propriedades tendem a estar associados a valores mais elevados de temperatura crítica.

Por outro lado, os atributos relacionados à valência média (wtd_mean_Valence, wtd_gmean_Valence e mean_Valence) apresentam correlação negativa moderada, indicando que, dentro deste conjunto de dados, materiais com maiores valores dessas variáveis tendem a apresentar menores temperaturas críticas.

Também é importante notar que nenhuma correlação se aproxima de ±1. A maior correlação observada está em torno de 0,7, o que indica que nenhum atributo individual é capaz de explicar sozinho a variabilidade da temperatura crítica. Isso sugere que o problema provavelmente depende da interação entre múltiplas características do material.

#### Relações individuais entre atributos e $T_c$

A correlação de Pearson pode não revelar associações não lineares. Por isso, construiremos gráficos de dispersão.

````python

atributos = list(top_atributos[:4])

fig, axes = plt.subplots(2, 2, figsize=(11, 8))

for ax, atributo in zip(axes.flat, atributos):
    ax.scatter(
        df[atributo],
        tc,
        alpha=0.12,
        s=10,
        color="steelblue",
        rasterized=True
    )

    ax.set_title(atributo)
    ax.set_xlabel(atributo)
    ax.set_ylabel("Temperatura crítica (K)")
    ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
````

Além da correlação observada, os gráficos de dispersão sugerem que a relação entre esses atributos e a temperatura crítica não é estritamente linear. 

Observam-se agrupamentos de materiais, regiões com diferentes níveis de dispersão e faixas discretas de valores, indicando que a temperatura crítica provavelmente depende de interações complexas entre múltiplas características físico-químicas. Esses resultados reforçam a hipótese de que modelos capazes de capturar relações não lineares, como Random Forest e XGBoost, podem ser mais adequados para a predição de $T_c$​ do que modelos lineares simples.

### Visualização gráfica e análise dos atributos

#### Investigando a redundância entre os atributos

Dois atributos podem conter informações semelhantes. Por exemplo, certas médias de propriedades elementares podem apresentar correlações elevadas entre si.

Isso é relevante porque variáveis redundantes podem afetar a interpretação dos modelos, aumentar sua complexidade e influenciar algoritmos 
sensíveis à multicolinearidade.

Vamos calcular a matriz de correlação apenas dos 81 preditores.

````python
X_eda = df.drop(columns=["critical_temp"])

correlacao = X_eda.corr(method="pearson")

fig, ax = plt.subplots(figsize=(9, 5))

im = ax.imshow(
    correlacao,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    aspect="auto"
)

fig.colorbar(im, ax=ax, label="Correlação de Pearson")

ax.set_title("Matriz de correlação entre os atributos")
ax.set_xlabel("Índice dos atributos")
ax.set_ylabel("Índice dos atributos")

plt.tight_layout()
plt.show()
````

A partir desse gráfico, podemos identificar os pares de atributos mais correlacionados.

````python

mascara = np.triu(
    np.ones(correlacao.shape, dtype=bool),
    k=1
)

pares = correlacao.where(mascara).stack()

pares_ordenados = (
    pares.reindex(
        pares.abs().sort_values(ascending=False).index
    )
)

display(
    pares_ordenados.head(15).to_frame(
        name="Correlação de Pearson"
    )
)
````

Observe se os pares mais correlacionados pertencem à mesma família de descritores químicos.

Não vamos remover automaticamente atributos muito correlacionados. Essa decisão fará parte da metodologia de seleção de atributos, no Capítulo 4. Por enquanto, estamos apenas identificando possíveis redundâncias.

#### Comparando as escalas dos atributos

A escala representa a ordem de grandeza dos valores de uma variável.

Por exemplo, um atributo relacionado à massa atômica pode apresentar valores muito diferentes de outro relacionado à energia de ionização.

Isso pode afetar especialmente o KNN, que utiliza distâncias entre observações, e a Ridge Regression, cuja regularização depende da escala dos coeficientes.

````python

estatisticas = X_eda.agg(["min", "max", "mean", "std"]).T

estatisticas["amplitude"] = (
    estatisticas["max"] - estatisticas["min"]
)

display(
    estatisticas.sort_values(
        "amplitude", ascending=False
    ).head(15)
)
````

Esse resultado nos ajudará a justificar posteriormente o uso de StandardScaler nos pipelines que necessitarem de padronização.

#### Verificando composições químicas repetidas

Segundo a documentação da UCI, o arquivo unique_m.csv contém as composições químicas correspondentes às observações do train.csv. 

Precisamos descobrir se diferentes registros representam a mesma composição química e se apresentam temperaturas críticas distintas.

Utilizaremos o método '.groupby' para reunir os valores repetidos e, assim, avaliá-los.

````python
repeticoes = (
    df_composicoes.groupby("material")
    .agg(
        quantidade=("critical_temp", "size"),
        tc_min=("critical_temp", "min"),
        tc_max=("critical_temp", "max"),
        tc_std=("critical_temp", "std")
    )
)

repeticoes["amplitude_tc"] = (
    repeticoes["tc_max"] - repeticoes["tc_min"]
)

repetidas = repeticoes[
    repeticoes["quantidade"] > 1
]

print("Composições repetidas:", len(repetidas))

display(
    repetidas.sort_values(
        "amplitude_tc", ascending=False
    ).head(15)
)
````

Se dois registros possuem a mesma composição, mas temperaturas críticas diferentes, quais informações ausentes poderiam explicar essa diferença?

Algumas hipóteses seriam diferenças estruturais, condições experimentais ou particularidades das amostras. Não poderemos identificar a causa sem informações adicionais.
